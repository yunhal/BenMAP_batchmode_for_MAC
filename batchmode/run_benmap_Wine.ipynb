```python
import os
import subprocess


def base_AQG_process_template(template_path, output_path, base_run, AQG_wine_dir, inmap_wine_dir, grid_level):
    """Reads a template file, replaces placeholders with actual paths, and saves the modified file."""
    with open(template_path, 'r') as f:
        content = f.read()
    
    # Replace placeholders
    content = content.replace('{AQG_wine_dir}', AQG_wine_dir)
    content = content.replace('{inmap_wine_dir}', inmap_wine_dir)
    content = content.replace('{base_run}', base_run)
    content = content.replace('{grid_level}', grid_level)

    print("AQG debug", output_path)
    # Write the modified content to a new file
    with open(output_path, 'w') as f:
        f.write(content)

def AQG_process_template(template_path, output_path, control_run, AQG_wine_dir, inmap_wine_dir, grid_level):
    """Reads a template file, replaces placeholders with actual paths, and saves the modified file."""
    with open(template_path, 'r') as f:
        content = f.read()
    
    # Replace placeholders
    content = content.replace('{AQG_wine_dir}', AQG_wine_dir)
    content = content.replace('{inmap_wine_dir}', inmap_wine_dir)
    content = content.replace('{control_run}', control_run)
    content = content.replace('{grid_level}', grid_level)

    print("AQG debug", output_path)
    # Write the modified content to a new file
    with open(output_path, 'w') as f:
        f.write(content)


def CFG_process_template(template_path, output_path, control_run, base_run, CFG_wine_dir, CFGR_wine_dir, AQG_wine_dir):
    """Reads a template file, replaces placeholders with actual paths, and saves the modified file."""
    with open(template_path, 'r') as f:
        content = f.read()
    
    # Replace placeholders
    content = content.replace('{CFG_wine_dir}', CFG_wine_dir)
    content = content.replace('{CFGR_wine_dir}', CFGR_wine_dir)
    content = content.replace('{AQG_wine_dir}', AQG_wine_dir)
    content = content.replace('{control_run}', control_run)
    content = content.replace('{base_run}', base_run)

    # Write the modified content to a new file
    with open(output_path, 'w') as f:
        f.write(content)

def APV_process_template(template_path, output_path, control_run, APV_wine_dir, APVR_wine_dir, CFGR_wine_dir, grid_level):
    """Reads a template file, replaces placeholders with actual paths, and saves the modified file."""
    with open(template_path, 'r') as f:
        content = f.read()
    
    # Replace placeholders
    content = content.replace('{APV_wine_dir}', APV_wine_dir)
    content = content.replace('{APVR_wine_dir}', APVR_wine_dir)
    content = content.replace('{CFGR_wine_dir}', CFGR_wine_dir)
    content = content.replace('{control_run}', control_run)
    content = content.replace('{grid_level}', grid_level)

    # Write the modified content to a new file
    with open(output_path, 'w') as f:
        f.write(content)

# Function to convert Unix paths to Wine-compatible Windows paths
def convert_to_wine_path(unix_path):
    windows_path = unix_path #s.replace('/', '\\')  # Convert to Windows-style backslashes
    return "Z:" + windows_path  # Prepend Z: for Wine


# Control run names

# control_runs = ['current_2020_NorthernGrid_West', 
#         #     'current_2020_SPP_North',
#         #     'current_2020_MISO_South',
#         #    'current_2020_CAISO', 
#         #    'current_2020_MISO_Central',
#            'current_2020_PJM_East',
#            'current_2020'
# ]


control_runs = ["2050_easyhard_decarb95","2050_easyhard_noIRA_111D"]  # "current_easyhard",
#control_runs = ["current_easyhard_Food_Agr","2050_easyhard_decarb95_Food_Agr","2050_easyhard_noIRA_111D_Food_Agr"] #  ['USA_CCS', 'USA_CCS_wo_NH3_VOC', 'CO_CCS', 'CO_CCS_wo_NH3_VOC', 'LA_CCS', 'LA_CCS_wo_NH3_VOC'] # ["CO_landfills_scenario1", "CO_landfills_scenario2" ] # "decarb95_2050_base","highREcost_2050_base"] # ,]#['ZERO'] #['CO_Suncor_CCS_wo_NH3_VOC', 'NEI_no_Landfill_2001411',  'CO_Cherokee_CCS_wo_NH3_VOC']
#run_name_list = ["current_2020","decarb95_2050","highREcost_2050", "current_2020_base","decarb95_2050_base","highREcost_2050_base" ]
are_multiple_base_runs = True  # egrid run has each base run for each control run 

# grid level
grid_dict =  {'County':'county'} # {'US Census Tracts' : 'tracts' } # 

# Key directories
Benmap_dir = '/Users/yunhalee/Documents/LOCAETA/RCM/BenMAP/'
AQG_dir = Benmap_dir + 'batchmode/AQG/'
CFG_dir = Benmap_dir +'batchmode/CFG/'
APV_dir = Benmap_dir +'batchmode/APV/'

# wine style directories
CFGR_wine_dir = 'Z:'+ Benmap_dir + 'batchmode/CFGR/'
APVR_wine_dir = 'Z:'+ Benmap_dir + 'batchmode/APVR/'

# Name of template ctlx file
ctlx_template = f'control_template_inmap_2020_pm25_wine.ctlx'
base_ctlx_template = f'base_template_inmap_2020_pm25_wine.ctlx'

# Wine executable path
benmap_exe = "C:\\Program Files (x86)\\BenMAP-CE\\BenMAP.exe"
for grid_level, grid_file_name in grid_dict.items():

    print (grid_level, grid_file_name)

    inmap_wine_dir = 'Z:' + Benmap_dir +'inmap_output/' + f'{grid_file_name}_case/' # directory of air quality input csv files

    # Base run placeholder in template
    if not are_multiple_base_runs: 
        #base_run = f'current_2020_base_{grid_file_name}_inmap_2020_pm25'
        base_run = f'base_nei2020_jun2025_{grid_file_name}_inmap_2020_pm25'

        # Process and run AQG template
        AQG_wine_dir = convert_to_wine_path(AQG_dir)
        aqg_template = os.path.join(AQG_dir, base_ctlx_template)
        aqg_output = os.path.join(AQG_dir, f'{base_run}_wine.ctlx')
        base_AQG_process_template(aqg_template, aqg_output, base_run, AQG_wine_dir, inmap_wine_dir, grid_level)
    
        subprocess.run(["wine", benmap_exe, convert_to_wine_path(aqg_output)], check=True)

    for run in control_runs:

        if are_multiple_base_runs:
            base_run = f'{run}_base_{grid_file_name}_inmap_2020_pm25'

            # Process and run AQG template
            AQG_wine_dir = convert_to_wine_path(AQG_dir)
            aqg_template = os.path.join(AQG_dir, base_ctlx_template)
            aqg_output = os.path.join(AQG_dir, f'{base_run}_wine.ctlx')
            base_AQG_process_template(aqg_template, aqg_output, base_run, AQG_wine_dir, inmap_wine_dir, grid_level)
            subprocess.run(["wine", benmap_exe, convert_to_wine_path(aqg_output)], check=True)

        control_run = f'control_{run}_{grid_file_name}_inmap_2020_pm25'
        #control_run = f'{run}_{grid_file_name}_inmap_2020_pm25'

        # Process and run AQG template
        AQG_wine_dir = convert_to_wine_path(AQG_dir)
        aqg_template = os.path.join(AQG_dir, ctlx_template)
        aqg_output = os.path.join(AQG_dir, f'{control_run}_wine.ctlx')
        AQG_process_template(aqg_template, aqg_output, control_run, AQG_wine_dir, inmap_wine_dir, grid_level)

        subprocess.run(["wine", benmap_exe, convert_to_wine_path(aqg_output)], check=True)

        # Process and run CFG template
        CFG_wine_dir = convert_to_wine_path(CFG_dir)
        cfg_template = os.path.join(CFG_dir, ctlx_template)
        cfg_output = os.path.join(CFG_dir, f'{control_run}_wine.ctlx')
        CFG_process_template(cfg_template, cfg_output, control_run, base_run, CFG_wine_dir, CFGR_wine_dir, AQG_wine_dir)

        subprocess.run(["wine", benmap_exe, convert_to_wine_path(cfg_output)], check=True)

        # Process and run APV template
        APV_wine_dir = convert_to_wine_path(APV_dir)
        apv_template = os.path.join(APV_dir, ctlx_template)
        apv_output = os.path.join(APV_dir, f'{control_run}_wine.ctlx')
        APV_process_template(apv_template, apv_output, control_run, APV_wine_dir, APVR_wine_dir, CFGR_wine_dir, grid_level)

        subprocess.run(["wine", benmap_exe, convert_to_wine_path(apv_output)], check=True)


```
